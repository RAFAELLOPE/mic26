### Data Loader for Nifti files

```python
from monai.transforms import (
    Compose, LoadImaged, EnsureChannelFirstd, Orientationd,
    NormalizeIntensityd, CropForegroundd, Resized, ToTensord, Lambdad
)
from monai.data import Dataset, DataLoader
import os
import nibabel as nib
import numpy as np
import torch
import matplotlib.pyplot as plt
```

```python
img_path = 'C:\\Users\\34616\\Documents\\MATLAB Drive\\neuro_db_T1_seg\\PD\\Sub-1524\\mwp11.3.46.670589.11.5320.5.0.4912.2017020109370757473.nii'
os.path.exists(img_path)
```

```python

loader = LoadImaged(keys=["image"])
img_t = loader({"image": img_path})
print(img_t["image"].shape)

plt.imshow(img_t["image"][50, :, :], cmap='gray')
```

```python
transforms = Compose([
    LoadImaged(keys=["image"]),
    EnsureChannelFirstd(keys=["image"]),
    Lambdad(keys=["image"], func=lambda x: torch.nan_to_num(x, nan=0.0)),

    Orientationd(keys=["image"], axcodes="RAS"),
    NormalizeIntensityd(keys=["image"]),
    Resized(keys=["image"], spatial_size=(128, 128, 128)),
    ToTensord(keys=["image", "label"]),
])


img_t = transforms({"image": img_path, "label":0})
img_t = img_t["image"].numpy()

print(img_t.shape)
plt.imshow(img_t[0, 50, :, :], cmap='gray')
```
